##Zomato data analysis

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("Zomato_data.csv")
```

#Convert the datatype of column-rate

```python
def handale_rate(value):
    value=str(value).split('/')
    value=value[0]
    return float(value)

df["rate"]=df["rate"].apply(handale_rate)
df.head()
```

Type of Restaurant

```python
sns.countplot(data=df,x="listed_in(type)")
plt.xlabel("Type of Restaurant")
plt.show()
```

Conclusion - Majority of the restaurant falls in dinning categorise.

```python
group_data=df.groupby('listed_in(type)')['votes'].sum()
result=pd.DataFrame({'votes': group_data})

plt.plot(result,c="green",marker="o")
plt.xlabel("type of restaurant", color="blue", fontsize=20)
plt.ylabel("votes", color="blue", fontsize=20)
plt.show()
```

Conclusion - Dinning restaurant has receiv maximum votes.

```python
plt.hist(df['rate'],bins=5)
plt.title("Rating distribution")
plt.show()
```

Conclusion - majority resturant receive rating from 3.5 to 4.

#Average order spending by couple

```python
couple_data=df['approx_cost(for two people)']
sns.countplot(x=couple_data)
plt.show()
```

Conclusion - The majority of couple prefer resturant with an approximate cost of 300 repees.

#Which mode receive maximum rating

```python
plt.figure(figsize=(6,6))
sns.boxplot(x="online_order",y="rate",data=df)
plt.show()
```

Conclusion - offline order receive lower rating as compare to online order.

```python
pivot_table=df.pivot_table(index="listed_in(type)",columns="online_order",aggfunc="size",fill_value=0)

sns.heatmap(pivot_table,annot=True)
plt.title("heatmap")
plt.xlabel("online_order")
plt.ylabel("listed_in(type)")
plt.show()
```

Conclusion - Dinning restaurant primarily accept offline orders, whereas cafes primarily receive online orders.This suggest that client preference order in a person at resturant,but prefer online ordering in cafes.
